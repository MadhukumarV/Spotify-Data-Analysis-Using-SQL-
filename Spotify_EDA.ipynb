```python
# Convert specific string column to numerical representation using Label Encoding

!pip install scikit-learn
import pandas as pd
from sklearn.preprocessing import LabelEncoder

# Assuming 'Artist' is the string column causing the issue
label_encoder = LabelEncoder()
df['Artist'] = label_encoder.fit_transform(df['Artist'])  # Encode 'Artist' column

# Now you can calculate the correlation using only numeric columns
corr_matrix = df.select_dtypes(include=['number']).corr()  # Select only numeric columns
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.show()
```
